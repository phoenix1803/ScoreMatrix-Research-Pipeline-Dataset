```python
# Cell 5: Define all metric functions
def preprocess_text(text):
    """Preprocess text for token-based metrics"""
    tokens = word_tokenize(str(text))
    lemmatizer = WordNetLemmatizer()
    return [lemmatizer.lemmatize(token.lower()) for token in tokens if token.lower() not in EN_STOPWORDS]

def exact_match(expected_answer, student_answer):
    """Exact string match"""
    return int(str(expected_answer) == str(student_answer))

def partial_match(expected_answer, student_answer):
    """Partial token-based matching"""
    expected_tokens = preprocess_text(expected_answer)
    student_tokens = preprocess_text(student_answer)
    common_tokens = set(expected_tokens) & set(student_tokens)
    return len(common_tokens) / max(len(expected_tokens), len(student_tokens)) if expected_tokens or student_tokens else 0

def cosine_similarity_score(expected_answer, student_answer):
    """TF-IDF based cosine similarity"""
    try:
        vectorizer = TfidfVectorizer(tokenizer=preprocess_text)
        tfidf_matrix = vectorizer.fit_transform([str(expected_answer), str(student_answer)])
        return float(cosine_similarity(tfidf_matrix[0], tfidf_matrix[1])[0][0])
    except:
        return 0.0

def sentiment_analysis(text):
    """Sentiment score using VADER"""
    try:
        sia = SentimentIntensityAnalyzer()
        return (sia.polarity_scores(str(text))['compound'] + 1) / 2
    except:
        return 0.5

def enhanced_sentence_match(expected_answer, student_answer):
    """Sentence transformer based semantic similarity"""
    try:
        embeddings = st_model.encode([str(expected_answer), str(student_answer)])
        return float(cosine_similarity([embeddings[0]], [embeddings[1]])[0][0])
    except:
        return 0.0

def multinomial_naive_bayes_score(expected_answer, student_answer):
    """Naive Bayes probability score"""
    try:
        answers = [str(expected_answer), str(student_answer)]
        vectorizer = CountVectorizer(tokenizer=preprocess_text)
        X = vectorizer.fit_transform(answers)
        y = [0, 1]
        clf = MultinomialNB()
        clf.fit(X, y)
        probs = clf.predict_proba(X)
        return float(probs[1][1])
    except:
        return 0.0

def semantic_similarity_score(expected_answer, student_answer):
    """Sentence transformer semantic similarity (same as enhanced_sentence_match)"""
    return enhanced_sentence_match(expected_answer, student_answer)

def coherence_score(text):
    """Text coherence based on vocabulary diversity"""
    try:
        sentences = sent_tokenize(str(text))
        if len(sentences) <= 1:
            return 1.0
        words = str(text).split()
        unique_words = len(set(words))
        total_words = len(words)
        return min(unique_words / max(1, total_words), 1.0)
    except:
        return 0.5

def relevance_score(expected_answer, student_answer):
    """Token overlap based relevance"""
    try:
        expected_tokens = set(preprocess_text(expected_answer))
        student_tokens = set(preprocess_text(student_answer))
        return len(expected_tokens & student_tokens) / len(expected_tokens) if expected_tokens else 0
    except:
        return 0.0

def grammatical_accuracy(text):
    """Grammar accuracy using LanguageTool"""
    try:
        matches = tool.check(str(text))
        words = str(text).split()
        return 1 - min(len(matches) / max(1, len(words)), 1.0)
    except:
        return 0.8

def keyword_coverage(expected_answer, student_answer):
    """Keyword coverage between answers"""
    try:
        expected_tokens = set(preprocess_text(expected_answer))
        student_tokens = set(preprocess_text(student_answer))
        return len(student_tokens & expected_tokens) / len(expected_tokens) if expected_tokens else 0
    except:
        return 0.0

def factual_consistency(expected_answer, student_answer):
    """NER-based factual consistency"""
    try:
        expected_doc = nlp(str(expected_answer))
        student_doc = nlp(str(student_answer))
        expected_facts = set([(ent.text, ent.label_) for ent in expected_doc.ents])
        student_facts = set([(ent.text, ent.label_) for ent in student_doc.ents])
        
        if len(expected_facts) == 0:
            return 1.0
        matching_facts = expected_facts & student_facts
        return len(matching_facts) / len(expected_facts)
    except:
        return 0.5

def structural_similarity(expected_answer, student_answer):
    """String sequence similarity"""
    try:
        return SequenceMatcher(None, str(expected_answer), str(student_answer)).ratio()
    except:
        return 0.0

def readability_score(text):
    """TextBlob-based readability"""
    try:
        analysis = TextBlob(str(text))
        polarity = (analysis.sentiment.polarity + 1) / 2
        subjectivity = 1 - analysis.sentiment.subjectivity
        return (polarity + subjectivity) / 2
    except:
        return 0.5

def concept_ordering(expected_answer, student_answer):
    """Word position ordering similarity"""
    try:
        expected_tokens = word_tokenize(str(expected_answer).lower())
        student_tokens = word_tokenize(str(student_answer).lower())
        expected_keywords = [word for word in expected_tokens if word not in EN_STOPWORDS]
        student_keywords = [word for word in student_tokens if word not in EN_STOPWORDS]
        common_keywords = set(expected_keywords) & set(student_keywords)
        
        if not common_keywords:
            return 0
            
        expected_positions = {word: i for i, word in enumerate(expected_keywords) if word in common_keywords}
        student_positions = {word: i for i, word in enumerate(student_keywords) if word in common_keywords}
        position_differences = sum(abs(expected_positions[word] - student_positions[word]) for word in common_keywords)
        max_possible_difference = len(expected_keywords) * len(student_keywords)
        
        if max_possible_difference == 0:
            return 1.0
        order_score = 1 - (position_differences / max_possible_difference)
        return max(0, order_score)
    except:
        return 0.0

def technical_accuracy(expected_answer, student_answer):
    """Technical term overlap using POS tagging"""
    try:
        expected_doc = nlp(str(expected_answer))
        student_doc = nlp(str(student_answer))
        expected_terms = set([token.text.lower() for token in expected_doc if token.pos_ in ['NOUN', 'PROPN']])
        student_terms = set([token.text.lower() for token in student_doc if token.pos_ in ['NOUN', 'PROPN']])
        
        if not expected_terms:
            return 1.0
        return len(expected_terms & student_terms) / len(expected_terms)
    except:
        return 0.5

def logical_flow(text):
    """Logical flow based on sentence structure"""
    try:
        sentences = sent_tokenize(str(text))
        if len(sentences) <= 1:
            return 0.5
        words = str(text).split()
        return min(1.0, (len(sentences) / max(1, len(words)) * 10))
    except:
        return 0.5

def weighted_average_score(scores, weights):
    """Calculate weighted average of scores"""
    total_weight = sum(weights.values())
    if total_weight == 0:
        return 0
    return sum(score * weights.get(criterion, 0) for criterion, score in scores.items()) / total_weight

print("All metric functions defined!")
```

```python
# Cell 7: Generate Gemini ground truth scores (respecting rate limits)
def gemini_score_batch(ref_answer, student_answer, max_marks=10):
    """Get Gemini score for a single answer pair"""
    prompt = f"""You are an expert teacher evaluating student answers.

Reference Answer: {ref_answer}
Student Answer: {student_answer}

Evaluate the student answer against the reference on a scale of 0-{max_marks}.
Consider: factual accuracy, completeness, clarity, and relevance.

Return ONLY a number between 0 and {max_marks}.
"""
    try:
        model = genai.GenerativeModel("gemini-1.5-flash-latest")
        response = model.generate_content(prompt)
        score = float(response.text.strip())
        return min(max(score, 0), max_marks)  # Clamp between 0 and max_marks
    except Exception as e:
        print(f"Gemini API error: {e}")
        return np.nan

# Generate Gemini scores with rate limiting
print("Generating Gemini ground truth scores...")
gemini_scores = []
for idx, row in sample_df.iterrows():
    score = gemini_score_batch(row["ref_answer"], row["student_answer"], 10)
    gemini_scores.append(score)
    print(f"Processed {len(gemini_scores)}/{len(sample_df)}: Score = {score}")
    time.sleep(4.5)  # Rate limiting for Gemini API

sample_df["gold_score_gemini"] = gemini_scores
sample_df.to_csv("03_sample_with_gemini.csv", index=False)
print(f"Generated {len([s for s in gemini_scores if not pd.isna(s)])} valid Gemini scores")
```

```python
# Cell 8: Compute all 17 metrics for the sample dataset
print("Computing all 17 metrics...")

# Initialize progress tracking
metrics_computed = []

def safe_apply(func, args):
    """Safely apply function and handle exceptions"""
    try:
        return func(*args)
    except Exception as e:
        return 0.0

# Compute all metrics
print("Computing exact_match...")
sample_df["exact_match"] = sample_df.apply(lambda r: safe_apply(exact_match, [r["ref_answer"], r["student_answer"]]), axis=1)

print("Computing partial_match...")
sample_df["partial_match"] = sample_df.apply(lambda r: safe_apply(partial_match, [r["ref_answer"], r["student_answer"]]), axis=1)

print("Computing cosine_similarity_new...")
sample_df["cosine_similarity_new"] = sample_df.apply(lambda r: safe_apply(cosine_similarity_score, [r["ref_answer"], r["student_answer"]]), axis=1)

print("Computing sentiment...")
sample_df["sentiment"] = sample_df["student_answer"].apply(lambda x: safe_apply(sentiment_analysis, [x]))

print("Computing enhanced_sentence_match...")
sample_df["enhanced_sentence_match"] = sample_df.apply(lambda r: safe_apply(enhanced_sentence_match, [r["ref_answer"], r["student_answer"]]), axis=1)

print("Computing multinomial_naive_bayes...")
sample_df["multinomial_naive_bayes"] = sample_df.apply(lambda r: safe_apply(multinomial_naive_bayes_score, [r["ref_answer"], r["student_answer"]]), axis=1)

print("Computing semantic_similarity...")
sample_df["semantic_similarity"] = sample_df.apply(lambda r: safe_apply(semantic_similarity_score, [r["ref_answer"], r["student_answer"]]), axis=1)

print("Computing coherence...")
sample_df["coherence"] = sample_df["student_answer"].apply(lambda x: safe_apply(coherence_score, [x]))

print("Computing relevance...")
sample_df["relevance"] = sample_df.apply(lambda r: safe_apply(relevance_score, [r["ref_answer"], r["student_answer"]]), axis=1)

print("Computing grammatical_accuracy...")
sample_df["grammatical_accuracy"] = sample_df["student_answer"].apply(lambda x: safe_apply(grammatical_accuracy, [x]))

print("Computing keyword_coverage...")
sample_df["keyword_coverage"] = sample_df.apply(lambda r: safe_apply(keyword_coverage, [r["ref_answer"], r["student_answer"]]), axis=1)

print("Computing factual_consistency...")
sample_df["factual_consistency"] = sample_df.apply(lambda r: safe_apply(factual_consistency, [r["ref_answer"], r["student_answer"]]), axis=1)

print("Computing structural_similarity...")
sample_df["structural_similarity"] = sample_df.apply(lambda r: safe_apply(structural_similarity, [r["ref_answer"], r["student_answer"]]), axis=1)

print("Computing readability...")
sample_df["readability"] = sample_df["student_answer"].apply(lambda x: safe_apply(readability_score, [x]))

print("Computing concept_ordering...")
sample_df["concept_ordering"] = sample_df.apply(lambda r: safe_apply(concept_ordering, [r["ref_answer"], r["student_answer"]]), axis=1)

print("Computing technical_accuracy...")
sample_df["technical_accuracy"] = sample_df.apply(lambda r: safe_apply(technical_accuracy, [r["ref_answer"], r["student_answer"]]), axis=1)

print("Computing logical_flow...")
sample_df["logical_flow"] = sample_df["student_answer"].apply(lambda x: safe_apply(logical_flow, [x]))

# Save intermediate results
sample_df.to_csv("04_sample_with_all_metrics.csv", index=False)
print("All 17 metrics computed successfully!")
```

```python
# Cell 9: Apply weighted scoring system 
weights = {
    'exact_match': 0.08,
    'partial_match': 0.06,
    'cosine_similarity_new': 0.06,  # Using new computed cosine similarity
    'sentiment': 0.04,
    'enhanced_sentence_match': 0.06,
    'multinomial_naive_bayes': 0.06,
    'semantic_similarity': 0.06,
    'coherence': 0.06,
    'relevance': 0.06,
    'grammatical_accuracy': 0.07,
    'keyword_coverage': 0.07,
    'factual_consistency': 0.08,
    'structural_similarity': 0.05,
    'readability': 0.05,
    'concept_ordering': 0.05,
    'technical_accuracy': 0.07,
    'logical_flow': 0.08
}

def compute_weighted_score(row, scale=10):
    """Compute weighted score normalized to 0-10 scale"""
    scores = {}
    for metric in weights.keys():
        if metric in row:
            scores[metric] = row[metric]
        else:
            scores[metric] = 0.0
    
    weighted_score = weighted_average_score(scores, weights)
    return weighted_score * scale

# Compute predicted scores
print("Computing weighted prediction scores...")
sample_df["pred_score"] = sample_df.apply(lambda row: compute_weighted_score(row, 10), axis=1)

# Also normalize original gold scores to 0-10 scale (originally 0-2)
sample_df["gold_score_normalized"] = sample_df["gold_score"] * (10/2)

sample_df.to_csv("05_sample_with_predictions.csv", index=False)
print("Weighted scoring complete!")
```
